## Creating a Tensor

### 1. Creating using Empty

```python
## Syntax --> .empty(row, column)
tensor = torch.empty(2,3)
```

```python
type(tensor)
```

### 2. Creating using Zeros and Ones

```python
## Syntax .zeros(rows, columns)
torch.zeros(2,3)
```

```python
## Similarly
torch.ones(3,4)
```

### 3. Creating using rand

```python
## Syntax .rand(rows, columns) --> Random value ranges between 0 and 1
torch.rand(5,6)
```

### 4. Creating using Seed

```python
torch.manual_seed(100)
torch.rand(5,6)
```

### 5. Creating using Tensors

```python
## Syntax -->torch.tensor(any data type)
torch.tensor([[1,2,3], [4,5,6]])
```

### 6. Other Ways

```python
# arange
print(f"Using arange function --> {torch.arange(0, 10)}")

# using linspace
# Syntax .linspace(start, stop, amount)
print(f"Using linspace function ---> {torch.linspace(0,10,10)}")

# using eye
print(f"Using eye function --->{'\n'} {torch.eye(5)}")

# using full
# Syntax .full((rows, columns), number)
print(f"Using full function -->{'\n'} {torch.full((3,3), 5)}")
```

## Mathematical Operations on Tensors

### 1. Scaler Operation

```python
x = torch.rand(2,2)
x
```

```python
## Addition
add = x + 2
print(f"Adding 2 to all values in a tensor {add}")

## Subtraction
sub = x - 2
print(f"Subtracting 2 to all values in a tensor {sub}")

## Multiplication
mul = x * 3
print(f"Multiplying 3 to all values in a tensor {mul}")

## Division
div = x / 5
print(f"Dividing 5 to all values in a tensor {div}")

## Int Division
a = x * 100
idiv = a // 3

## Mod
mod = idiv % 2

## Power
pow = x ** 2
print(f"Exponent by 2 {pow}")
```

```python
a = torch.rand(2,3)
b = torch.rand(2,3)

print(a)
print(b)
```

```python
## Add
a + b

## Sub
a - b

## Multiply
a * b

## Division
a / b

## Power
a ** b
```

### 2. Reduction Operation

```python
e = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float32)
e
```

```python
## sum
torch.sum(e)
```

```python
## Sum along columns
torch.sum(e, dim=0)
```

```python
## Sum along rows
torch.sum(e, dim=1)
```

```python
## Mean
torch.mean(e, dtype=torch.float32)
```

```python
## Mean along columns
torch.mean(e, dtype=torch.float32, dim=0)
```

```python
## Median
torch.median(e)
```

```python
## Max and Min
torch.max(e)
```

```python
torch.min(e)
```

```python
## Product
torch.prod(e)
```

```python
## STD
torch.std(e)
```

```python
## Variance
torch.var(e)
```

```python
## Argmax
torch.argmax(e)
```

```python
torch.argmin(e)
```
